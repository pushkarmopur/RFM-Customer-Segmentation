# Notebook 03 — Final K-Means K=5 and Customer Confidence

## Purpose

Notebook 02 identified **K=2 as the broad statistical benchmark** and carried **K=5 forward as the operational candidate** because it provided greater customer-behaviour granularity while retaining acceptable validation, strong stability and practical cluster sizes.

This notebook now freezes **K-Means K=5** as the operational segmentation and completes the final K-Means analysis. It:

1. validates the same 4,338-customer, seven-feature modelling population;
2. fits and saves the final K-Means K=5 model;
3. calculates the final internal validation metrics;
4. creates original-scale cluster profiles and business segment names;
5. quantifies customer and Monetary contribution by segment;
6. measures **customer-level assignment confidence** using individual Silhouette values and relative centroid margins;
7. records each customer's second-nearest K-Means segment;
8. separates strong, moderate and boundary/review assignments;
9. independently re-tests the frozen model using 30 random seeds and 50 repeated 80% customer subsamples;
10. prepares final K-Means outputs for later GMM comparison and dashboard construction.

> **Interpretation rule:** customer-confidence categories describe how clearly a customer fits the fitted K-Means geometry. They are **not classification accuracy or misclassification rates**, because no ground-truth customer segment labels exist.


## Phase 1 — Setup and input validation

The final K=5 model must use exactly the same customer population and seven processed behavioural features established in Notebook 01 and evaluated in Notebook 02.


### 1. Import the required libraries




```python
import os
import sys
import json
import shutil
import platform
import warnings
import joblib
from pathlib import Path
from importlib import metadata

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn

from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import (
    silhouette_score,
    silhouette_samples,
    davies_bouldin_score,
    calinski_harabasz_score,
    adjusted_rand_score
)

from IPython.display import display

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 200)
pd.set_option("display.float_format", lambda value: f"{value:,.4f}")

sns.set_style("whitegrid")

print("Libraries imported successfully.")
```

### 2. Define the Notebook 01 inputs and Notebook 03 output folders

The notebook accepts either the two required CSV files directly in `/content` or the ZIP produced by Notebook 01. When the ZIP is available, the required processed dataset, original-scale customer dataset and fitted RobustScaler are located automatically.


```python
# ------------------------------------------------------------
# Notebook 01 handoff
# ------------------------------------------------------------

NOTEBOOK01_ZIP = Path(
    "/content/01_Data_Preparation_and_Feature_Engineering_All_Results.zip"
)
NOTEBOOK01_EXTRACT_DIR = Path("/content/Notebook01_Extracted")

DIRECT_PROCESSED_FILE = Path("/content/customer_features_processed.csv")
DIRECT_RAW_FILE = Path("/content/customer_features_ready_for_preprocessing.csv")

def find_file_recursive(root, filename):
    matches = list(Path(root).rglob(filename))
    return matches[0] if matches else None

if DIRECT_PROCESSED_FILE.exists() and DIRECT_RAW_FILE.exists():
    processed_file = str(DIRECT_PROCESSED_FILE)
    raw_file = str(DIRECT_RAW_FILE)
else:
    if NOTEBOOK01_ZIP.exists():
        if NOTEBOOK01_EXTRACT_DIR.exists():
            shutil.rmtree(NOTEBOOK01_EXTRACT_DIR)
        NOTEBOOK01_EXTRACT_DIR.mkdir(parents=True, exist_ok=True)
        shutil.unpack_archive(
            str(NOTEBOOK01_ZIP),
            str(NOTEBOOK01_EXTRACT_DIR)
        )

    processed_match = find_file_recursive(
        "/content",
        "customer_features_processed.csv"
    )
    raw_match = find_file_recursive(
        "/content",
        "customer_features_ready_for_preprocessing.csv"
    )

    if processed_match is None or raw_match is None:
        raise FileNotFoundError(
            "Notebook 03 could not locate the two required Notebook 01 outputs. "
            "Upload either both CSV files or "
            "01_Data_Preparation_and_Feature_Engineering_All_Results.zip to /content."
        )

    processed_file = str(processed_match)
    raw_file = str(raw_match)

# ------------------------------------------------------------
# Notebook 03 result folders
# ------------------------------------------------------------

results_dir = Path("/content/03_Final_KMeans_K5_and_Confidence_Results")
tables_dir = results_dir / "Tables"
figures_dir = results_dir / "Figures"
models_dir = results_dir / "Models"
datasets_dir = results_dir / "Datasets"
stability_dir = results_dir / "Stability"
reproducibility_dir = results_dir / "Reproducibility"

for folder in [
    results_dir, tables_dir, figures_dir, models_dir,
    datasets_dir, stability_dir, reproducibility_dir
]:
    folder.mkdir(parents=True, exist_ok=True)

print("Processed file:", processed_file)
print("Original-scale file:", raw_file)
print("Output folder:", results_dir)
```

### 3. Load the two customer datasets


```python
if not os.path.exists(processed_file):
    raise FileNotFoundError(
        f"Processed dataset not found: {processed_file}"
    )

if not os.path.exists(raw_file):
    raise FileNotFoundError(
        "Original-scale customer feature dataset was not found. "
        "Upload customer_features_ready_for_preprocessing.csv."
    )

processed_data = pd.read_csv(processed_file)
raw_data = pd.read_csv(raw_file)

print("=" * 80)
print("DATASETS LOADED")
print("=" * 80)

print("\nProcessed dataset shape:", processed_data.shape)
print("Original-scale dataset shape:", raw_data.shape)

print("\nProcessed dataset preview:")
display(processed_data.head())

print("\nOriginal-scale dataset preview:")
display(raw_data.head())
```

### 4. Validate columns, missing values and customer IDs


```python
feature_columns = [
    "Recency",
    "Frequency",
    "Monetary",
    "AverageOrderValue",
    "BasketSize",
    "ProductDiversity",
    "TotalQuantity"
]

required_columns = ["CustomerID"] + feature_columns

missing_processed = [
    column for column in required_columns
    if column not in processed_data.columns
]

missing_raw = [
    column for column in required_columns
    if column not in raw_data.columns
]

if missing_processed:
    raise ValueError(
        f"Missing columns in processed dataset: {missing_processed}"
    )

if missing_raw:
    raise ValueError(
        f"Missing columns in original-scale dataset: {missing_raw}"
    )

if processed_data["CustomerID"].duplicated().any():
    raise ValueError(
        "Duplicate CustomerID values found in processed dataset."
    )

if raw_data["CustomerID"].duplicated().any():
    raise ValueError(
        "Duplicate CustomerID values found in original-scale dataset."
    )

if processed_data[required_columns].isnull().sum().sum() > 0:
    raise ValueError(
        "Missing values found in processed clustering dataset."
    )

if raw_data[required_columns].isnull().sum().sum() > 0:
    raise ValueError(
        "Missing values found in original-scale dataset."
    )

print("Required columns and customer IDs validated successfully.")
```

### 5. Align the datasets and create the feature matrix


```python
raw_data = (
    processed_data[["CustomerID"]]
    .merge(
        raw_data[required_columns],
        on="CustomerID",
        how="left",
        validate="one_to_one"
    )
)

if raw_data.isnull().sum().sum() > 0:
    raise ValueError(
        "CustomerID alignment failed between the two datasets."
    )

if not np.array_equal(
    processed_data["CustomerID"].to_numpy(),
    raw_data["CustomerID"].to_numpy()
):
    raise ValueError(
        "Customer ordering is not aligned between the datasets."
    )

X = processed_data[feature_columns].copy()

if not np.isfinite(X.to_numpy()).all():
    raise ValueError(
        "Infinite or non-finite values found in clustering features."
    )

print("=" * 80)
print("INPUT VALIDATION COMPLETED")
print("=" * 80)

print("\nCustomers:", len(X))
print("Features:", len(feature_columns))
print("\nFeatures used:")
print(feature_columns)

display(X.describe().T)
```

## Phase 2 — Freeze K-Means K=5 and calculate internal validation

The final K=5 model is fitted and its compactness, separation and cluster sizes are examined.


### 6. Train the final K=5 model

K-Means is fitted using K-Means++ initialisation, 50 initialisations, 500 maximum iterations and a fixed random state. Labels are added to both datasets and the model is saved.


```python
FINAL_K = 5
RANDOM_STATE = 42
N_INIT = 50
MAX_ITER = 500

final_kmeans = KMeans(
    n_clusters=FINAL_K,
    init="k-means++",
    n_init=N_INIT,
    max_iter=MAX_ITER,
    random_state=RANDOM_STATE,
    algorithm="lloyd"
)

final_labels = final_kmeans.fit_predict(X)

processed_clustered = processed_data.copy()
processed_clustered["Cluster"] = final_labels

raw_clustered = raw_data.copy()
raw_clustered["Cluster"] = final_labels

joblib.dump(
    final_kmeans,
    models_dir / "final_kmeans_k5_model.joblib"
)

with open(
    models_dir / "final_feature_columns.json",
    "w"
) as file:
    json.dump(feature_columns, file, indent=4)

print("=" * 80)
print("FINAL K-MEANS MODEL TRAINED")
print("=" * 80)

print("Selected K:", FINAL_K)
print("Iterations to converge:", final_kmeans.n_iter_)
print("Cluster labels:", sorted(np.unique(final_labels)))
```

### 7. Reload and verify the saved model




```python
# VERIFY THE SAVED K-MEANS MODEL

saved_model_path = (
    models_dir / "final_kmeans_k5_model.joblib"
)

reloaded_kmeans = joblib.load(
    saved_model_path
)

reloaded_labels = reloaded_kmeans.predict(X)

if not np.array_equal(
    final_labels,
    reloaded_labels
):
    raise AssertionError(
        "The reloaded K-Means model did not reproduce "
        "the original cluster assignments."
    )

print(
    "Saved model verification passed: "
    "all cluster assignments were reproduced."
)
```

### 8. Calculate the final validation metrics
Inertia, Silhouette Score, Davies-Bouldin Index and Calinski-Harabasz Index are calculated and saved.



```python
final_silhouette = silhouette_score(X, final_labels)
final_dbi = davies_bouldin_score(X, final_labels)
final_ch = calinski_harabasz_score(X, final_labels)
final_inertia = final_kmeans.inertia_

final_validation = pd.DataFrame({
    "Metric": [
        "Selected K",
        "WCSS / Inertia",
        "Silhouette Score",
        "Davies-Bouldin Index",
        "Calinski-Harabasz Index",
        "Iterations to Converge"
    ],
    "Value": [
        FINAL_K,
        final_inertia,
        final_silhouette,
        final_dbi,
        final_ch,
        final_kmeans.n_iter_
    ]
})

print("=" * 80)
print("FINAL INTERNAL VALIDATION")
print("=" * 80)

display(final_validation)

final_validation.to_csv(
    tables_dir / "final_kmeans_validation_metrics.csv",
    index=False
)
```

### 9. Calculate customer count in each cluster



```python
cluster_size = (
    raw_clustered["Cluster"]
    .value_counts()
    .sort_index()
    .reindex(range(FINAL_K), fill_value=0)
    .rename_axis("Cluster")
    .reset_index(name="Customer_Count")
)

cluster_size["Customer_Percentage"] = (
    cluster_size["Customer_Count"]
    / len(raw_clustered)
    * 100
).round(2)

cluster_size["Is_Smaller_Than_5_Percent"] = (
    cluster_size["Customer_Percentage"] < 5
)

print("=" * 80)
print("FINAL CLUSTER SIZE ANALYSIS")
print("=" * 80)

display(cluster_size)

cluster_size.to_csv(
    tables_dir / "final_cluster_size_analysis.csv",
    index=False
)
```

### 10. Plot the cluster sizes




```python
plt.figure(figsize=(9, 5))

ax = sns.barplot(
    data=cluster_size,
    x="Cluster",
    y="Customer_Count",
    color="#4C72B0"
)

for index, row in cluster_size.iterrows():
    ax.text(
        index,
        row["Customer_Count"],
        (
            f'{row["Customer_Count"]:,}\n'
            f'({row["Customer_Percentage"]:.1f}%)'
        ),
        ha="center",
        va="bottom"
    )

plt.title("Final K-Means Cluster Sizes — K = 5")
plt.xlabel("Cluster")
plt.ylabel("Number of Customers")
plt.tight_layout()

plt.savefig(
    figures_dir / "final_cluster_sizes_by_number.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### 11. Calculate customer and cluster silhouette values



```python
customer_silhouette = silhouette_samples(X, final_labels)

processed_clustered["Silhouette_Value"] = customer_silhouette
raw_clustered["Silhouette_Value"] = customer_silhouette

silhouette_summary = (
    raw_clustered
    .groupby("Cluster")["Silhouette_Value"]
    .agg(
        Customer_Count="count",
        Mean_Silhouette="mean",
        Median_Silhouette="median",
        Minimum_Silhouette="min",
        Maximum_Silhouette="max",
        Standard_Deviation="std"
    )
    .reset_index()
)

negative_counts = (
    raw_clustered
    .assign(Is_Negative=raw_clustered["Silhouette_Value"] < 0)
    .groupby("Cluster")["Is_Negative"]
    .sum()
    .reset_index(name="Negative_Silhouette_Customers")
)

silhouette_summary = silhouette_summary.merge(
    negative_counts,
    on="Cluster",
    how="left"
)

silhouette_summary["Negative_Silhouette_Percentage"] = (
    silhouette_summary["Negative_Silhouette_Customers"]
    / silhouette_summary["Customer_Count"]
    * 100
).round(2)

silhouette_summary = silhouette_summary.round(4)

overall_negative_count = int(
    (customer_silhouette < 0).sum()
)

overall_negative_percentage = (
    overall_negative_count
    / len(customer_silhouette)
    * 100
)

overall_silhouette_summary = pd.DataFrame({
    "Measure": [
        "Overall Silhouette Score",
        "Ambiguous / Boundary Customers",
        "Ambiguous / Boundary Customer Percentage"
    ],
    "Value": [
        final_silhouette,
        overall_negative_count,
        overall_negative_percentage
    ]
})

print("=" * 80)
print("SILHOUETTE QUALITY BY CLUSTER")
print("=" * 80)

display(silhouette_summary)

print("\nOverall silhouette summary:")
display(overall_silhouette_summary)

silhouette_summary.to_csv(
    tables_dir / "final_cluster_silhouette_summary.csv",
    index=False
)

overall_silhouette_summary.to_csv(
    tables_dir / "overall_silhouette_summary.csv",
    index=False
)
```

### 12. Plot the silhouette distribution



```python
plt.figure(figsize=(10, 7))

y_lower = 10

for cluster_number in range(FINAL_K):
    cluster_values = customer_silhouette[
        final_labels == cluster_number
    ]

    cluster_values.sort()

    cluster_count = cluster_values.shape[0]
    y_upper = y_lower + cluster_count

    plt.fill_betweenx(
        np.arange(y_lower, y_upper),
        0,
        cluster_values,
        alpha=0.70
    )

    plt.text(
        -0.06,
        y_lower + 0.5 * cluster_count,
        str(cluster_number)
    )

    y_lower = y_upper + 10

plt.axvline(
    final_silhouette,
    linestyle="--",
    label=f"Average Silhouette = {final_silhouette:.4f}"
)

plt.title("Silhouette Plot for Final K-Means Model — K = 5")
plt.xlabel("Silhouette Coefficient")
plt.ylabel("Cluster")
plt.legend()
plt.tight_layout()

plt.savefig(
    figures_dir / "final_silhouette_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Phase 3 — Original-scale profiling and business interpretation

The original customer values are used to understand, name and compare the five groups.


### 13. Create mean and median cluster profiles


```python
cluster_profile_mean = (
    raw_clustered
    .groupby("Cluster")[feature_columns]
    .mean()
    .round(2)
)

cluster_profile_median = (
    raw_clustered
    .groupby("Cluster")[feature_columns]
    .median()
    .round(2)
)

cluster_profile_mean["Customer_Count"] = (
    raw_clustered.groupby("Cluster").size()
)

cluster_profile_mean["Customer_Percentage"] = (
    cluster_profile_mean["Customer_Count"]
    / len(raw_clustered)
    * 100
).round(2)

cluster_profile_median["Customer_Count"] = (
    raw_clustered.groupby("Cluster").size()
)

cluster_profile_median["Customer_Percentage"] = (
    cluster_profile_median["Customer_Count"]
    / len(raw_clustered)
    * 100
).round(2)

print("=" * 80)
print("ORIGINAL-SCALE MEAN PROFILES")
print("=" * 80)
display(cluster_profile_mean)

print("\nOriginal-scale median profiles:")
display(cluster_profile_median)

cluster_profile_mean.to_csv(
    tables_dir / "final_cluster_profiles_mean.csv"
)

cluster_profile_median.to_csv(
    tables_dir / "final_cluster_profiles_median.csv"
)
```

### 14. Compare mean and median Monetary values



```python
mean_median_comparison = pd.DataFrame({
    "Cluster": cluster_profile_mean.index,
    "Mean_Monetary": cluster_profile_mean["Monetary"].values,
    "Median_Monetary": cluster_profile_median["Monetary"].values
})

mean_median_comparison["Mean_to_Median_Ratio"] = (
    mean_median_comparison["Mean_Monetary"]
    / mean_median_comparison["Median_Monetary"].replace(0, np.nan)
)

mean_median_comparison["Absolute_Difference"] = (
    mean_median_comparison["Mean_Monetary"]
    - mean_median_comparison["Median_Monetary"]
)

mean_median_comparison = mean_median_comparison.round(2)

display(mean_median_comparison)

mean_median_comparison.to_csv(
    tables_dir / "mean_vs_median_monetary_comparison.csv",
    index=False
)
```

### 15. Create relative cluster profile scores



```python
profile_values = cluster_profile_mean[feature_columns].copy()

relative_profiles = (
    profile_values - profile_values.mean()
) / profile_values.std(ddof=0).replace(0, 1)

relative_profiles = relative_profiles.round(3)

relative_profiles.to_csv(
    tables_dir / "final_cluster_relative_profile_scores.csv"
)

plt.figure(figsize=(12, 6))

sns.heatmap(
    relative_profiles,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Relative Customer Profiles — Final K = 5")
plt.xlabel("Customer Behaviour Features")
plt.ylabel("Cluster")
plt.tight_layout()

plt.savefig(
    figures_dir / "final_relative_profile_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### 16. Derive the five business segment names




```python
profile_for_naming = cluster_profile_mean[feature_columns].copy()

remaining_clusters = set(profile_for_naming.index)
segment_mapping = {}

# Most dormant group: highest average Recency
dormant_cluster = profile_for_naming.loc[
    list(remaining_clusters),
    "Recency"
].idxmax()

segment_mapping[int(dormant_cluster)] = (
    "Dormant Occasional Customers"
)

remaining_clusters.remove(dormant_cluster)

# Most loyal group: highest Frequency
loyal_cluster = profile_for_naming.loc[
    list(remaining_clusters),
    "Frequency"
].idxmax()

segment_mapping[int(loyal_cluster)] = (
    "High-Value Loyal Customers"
)

remaining_clusters.remove(loyal_cluster)

# Large-order group: highest Average Order Value
large_order_cluster = profile_for_naming.loc[
    list(remaining_clusters),
    "AverageOrderValue"
].idxmax()

segment_mapping[int(large_order_cluster)] = (
    "High-Value Large-Order Buyers"
)

remaining_clusters.remove(large_order_cluster)

# Lowest-value group: lowest Monetary value
low_value_cluster = profile_for_naming.loc[
    list(remaining_clusters),
    "Monetary"
].idxmin()

segment_mapping[int(low_value_cluster)] = (
    "Low-Value Infrequent Customers"
)

remaining_clusters.remove(low_value_cluster)

# Remaining middle-value group
regular_cluster = list(remaining_clusters)[0]

segment_mapping[int(regular_cluster)] = (
    "Active Regular Customers"
)

segment_mapping_table = pd.DataFrame({
    "Cluster": sorted(segment_mapping),
    "Segment_Name": [
        segment_mapping[cluster]
        for cluster in sorted(segment_mapping)
    ]
})

if segment_mapping_table["Segment_Name"].nunique() != FINAL_K:
    raise ValueError(
        "Segment naming did not produce five unique business names."
    )

print("=" * 80)
print("FINAL BUSINESS SEGMENT MAPPING")
print("=" * 80)

display(segment_mapping_table)

segment_mapping_table.to_csv(
    tables_dir / "final_segment_mapping.csv",
    index=False
)
```

### 17. Apply the business names to the result tables

 The same business label must be attached consistently to every customer and summary table.So,the segment mapping is applied to the processed data, original-scale data, cluster-size table, silhouette table and profile tables.



```python
processed_clustered["Segment_Name"] = (
    processed_clustered["Cluster"].map(segment_mapping)
)

raw_clustered["Segment_Name"] = (
    raw_clustered["Cluster"].map(segment_mapping)
)

cluster_size["Segment_Name"] = (
    cluster_size["Cluster"].map(segment_mapping)
)

silhouette_summary["Segment_Name"] = (
    silhouette_summary["Cluster"].map(segment_mapping)
)

cluster_profile_mean_named = (
    cluster_profile_mean.reset_index()
)

cluster_profile_mean_named["Segment_Name"] = (
    cluster_profile_mean_named["Cluster"].map(segment_mapping)
)

cluster_profile_median_named = (
    cluster_profile_median.reset_index()
)

cluster_profile_median_named["Segment_Name"] = (
    cluster_profile_median_named["Cluster"].map(segment_mapping)
)

display(
    cluster_profile_mean_named[
        [
            "Cluster",
            "Segment_Name",
            *feature_columns,
            "Customer_Count",
            "Customer_Percentage"
        ]
    ]
)
```

### 18. Plot customer distribution by segment name




```python
named_cluster_size = cluster_size.copy()

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=named_cluster_size,
    x="Segment_Name",
    y="Customer_Count",
    color="#4C72B0"
)

for index, row in named_cluster_size.iterrows():
    ax.text(
        index,
        row["Customer_Count"],
        (
            f'{row["Customer_Count"]:,}\n'
            f'({row["Customer_Percentage"]:.1f}%)'
        ),
        ha="center",
        va="bottom"
    )

plt.title("Customer Distribution Across Final Segments")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()

plt.savefig(
    figures_dir / "final_named_segment_sizes.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### 19. Calculate Monetary contribution by segment



```python
segment_value_summary = (
    raw_clustered
    .groupby(
        ["Cluster", "Segment_Name"]
    )
    .agg(
        Customer_Count=("CustomerID", "count"),
        Total_Monetary_Value=("Monetary", "sum"),
        Mean_Customer_Value=("Monetary", "mean"),
        Median_Customer_Value=("Monetary", "median")
    )
    .reset_index()
)

segment_value_summary["Customer_Percentage"] = (
    segment_value_summary["Customer_Count"]
    / len(raw_clustered)
    * 100
)

segment_value_summary["Monetary_Value_Percentage"] = (
    segment_value_summary["Total_Monetary_Value"]
    / segment_value_summary["Total_Monetary_Value"].sum()
    * 100
)

segment_value_summary = segment_value_summary.round(2)

print("=" * 80)
print("MONETARY CONTRIBUTION BY SEGMENT")
print("=" * 80)

display(segment_value_summary)

segment_value_summary.to_csv(
    tables_dir / "final_segment_monetary_contribution.csv",
    index=False
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=segment_value_summary,
    x="Segment_Name",
    y="Monetary_Value_Percentage",
    color="#4C72B0"
)

plt.title("Monetary Value Contribution by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Percentage of Total Monetary Value")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()

plt.savefig(
    figures_dir / "segment_monetary_contribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### 20. Create segment feature quartiles

## Phase 4 — Customer-level assignment confidence and boundary analysis

### Purpose

K-Means is an unsupervised clustering algorithm. There are no known correct
customer segment labels, so traditional classification accuracy,
misclassification rate, precision, recall and a confusion matrix cannot be
calculated.

This phase instead evaluates **how clearly each customer fits the assigned
cluster**.

Two complementary measures are used:

1. **Individual Silhouette Value**
   - positive: the customer is generally closer to the assigned cluster;
   - near zero: the customer lies close to a cluster boundary;
   - negative: the customer is, on average, closer to a neighbouring cluster.

2. **Relative Centroid Margin**
   - compares the distance to the assigned centroid with the distance to the
     second-nearest centroid;
   - a smaller margin indicates a less decisive K-Means assignment.

The analysis also records the customer's second-closest cluster and business
segment. This provides an interpretable alternative segment for customers
positioned near a boundary.


### 23. Calculate the nearest and second-nearest centroid distances

For customer $i$, let $d_{1,i}$ be the distance to the assigned centroid and $d_{2,i}$ the distance to the second-nearest centroid. The relative centroid margin is calculated as:

$$\text{Relative Centroid Margin}_i = \frac{d_{2,i}-d_{1,i}}{d_{2,i}}$$

- A value close to **0** means the two nearest centroids are almost equally distant, so the assignment is less decisive.
- A larger value means the assigned centroid has a clearer distance advantage over the alternative.

The second-nearest cluster is also retained as the customer's most plausible alternative K-Means segment.


```python
# CUSTOMER-LEVEL CENTROID DISTANCE ANALYSIS

centroid_distances = final_kmeans.transform(X)

ordered_cluster_indices = np.argsort(
    centroid_distances,
    axis=1
)

nearest_cluster_from_distance = (
    ordered_cluster_indices[:, 0]
)

second_nearest_cluster = (
    ordered_cluster_indices[:, 1]
)

# The nearest-centroid result must reproduce the fitted K-Means labels.
assert np.array_equal(
    nearest_cluster_from_distance,
    final_labels
), (
    "Nearest-centroid assignments did not reproduce "
    "the fitted K-Means labels."
)

row_indices = np.arange(len(X))

assigned_centroid_distance = centroid_distances[
    row_indices,
    final_labels
]

second_centroid_distance = centroid_distances[
    row_indices,
    second_nearest_cluster
]

absolute_centroid_gap = (
    second_centroid_distance
    - assigned_centroid_distance
)

safe_second_distance = np.maximum(
    second_centroid_distance,
    np.finfo(float).eps
)

relative_centroid_margin = (
    absolute_centroid_gap
    / safe_second_distance
)

assigned_to_second_distance_ratio = (
    assigned_centroid_distance
    / safe_second_distance
)

centroid_margin_summary = pd.Series(
    relative_centroid_margin,
    name="Relative_Centroid_Margin"
).describe(
    percentiles=[
        0.01,
        0.05,
        0.10,
        0.25,
        0.50,
        0.75,
        0.90,
        0.95,
        0.99
    ]
)

print("=" * 80)
print("CENTROID DISTANCE ANALYSIS")
print("=" * 80)

display(
    centroid_margin_summary.to_frame()
)
```

### 24. Define transparent customer-confidence conditions

The final confidence framework combines two forms of geometric evidence.

**Boundary / review assignment**

A customer is marked for additional review when:

$$\text{Silhouette}_i < 0 \quad \text{OR} \quad \text{Relative Centroid Margin}_i < 0.10$$

**Moderate assignment**

The customer is not in the review group, but either the individual Silhouette is below 0.25 or the relative centroid margin is below 0.30.

**Strong assignment**

The customer has an individual Silhouette of at least 0.25 **and** a relative centroid margin of at least 0.30.

These thresholds are transparent analytical rules defined for this dissertation. They are not universal K-Means probabilities or supervised classification thresholds. A threshold-sensitivity table is produced later in the notebook.


```python
# DEFINE CUSTOMER-LEVEL CONFIDENCE CONDITIONS

BOUNDARY_MARGIN_THRESHOLD = 0.10
MODERATE_SILHOUETTE_THRESHOLD = 0.25
MODERATE_MARGIN_THRESHOLD = 0.30

negative_silhouette_flag = (
    customer_silhouette < 0
)

low_centroid_margin_flag = (
    relative_centroid_margin
    < BOUNDARY_MARGIN_THRESHOLD
)

combined_boundary_flag = (
    negative_silhouette_flag
    | low_centroid_margin_flag
)

assignment_confidence_category = np.select(
    [
        combined_boundary_flag,
        (
            (
                customer_silhouette
                < MODERATE_SILHOUETTE_THRESHOLD
            )
            |
            (
                relative_centroid_margin
                < MODERATE_MARGIN_THRESHOLD
            )
        )
    ],
    [
        "Boundary / Ambiguous Assignment",
        "Moderate Assignment"
    ],
    default="Strong Assignment"
)

alternative_segment = pd.Series(
    second_nearest_cluster
).map(
    segment_mapping
).to_numpy()

confidence_fields = {
    "Alternative_Cluster": second_nearest_cluster,
    "Alternative_Segment": alternative_segment,
    "Assigned_Centroid_Distance": (
        assigned_centroid_distance
    ),
    "Second_Centroid_Distance": (
        second_centroid_distance
    ),
    "Absolute_Centroid_Distance_Gap": (
        absolute_centroid_gap
    ),
    "Relative_Centroid_Margin": (
        relative_centroid_margin
    ),
    "Assigned_to_Second_Distance_Ratio": (
        assigned_to_second_distance_ratio
    ),
    "Negative_Silhouette_Flag": (
        negative_silhouette_flag
    ),
    "Low_Centroid_Margin_Flag": (
        low_centroid_margin_flag
    ),
    "Combined_Boundary_Flag": (
        combined_boundary_flag
    ),
    "Assignment_Confidence_Category": (
        assignment_confidence_category
    )
}

for column_name, column_values in confidence_fields.items():
    processed_clustered[column_name] = column_values
    raw_clustered[column_name] = column_values

print(
    "Customer-confidence columns added to the "
    "processed and original-scale customer datasets."
)
```

### 25. Summarise confidence across all customers and within each segment

The summary distinguishes:

- customers with negative Silhouette Values;
- customers with a low centroid margin;
- customers satisfying both conditions;
- the combined boundary group;
- moderate assignments;
- strong assignments.

The segment-level results identify where assignment uncertainty is
concentrated.


```python
# OVERALL CUSTOMER-CONFIDENCE SUMMARY

negative_silhouette_count = int(
    negative_silhouette_flag.sum()
)

low_centroid_margin_count = int(
    low_centroid_margin_flag.sum()
)

both_boundary_conditions_count = int(
    (
        negative_silhouette_flag
        & low_centroid_margin_flag
    ).sum()
)

combined_boundary_count = int(
    combined_boundary_flag.sum()
)

moderate_assignment_count = int(
    (
        assignment_confidence_category
        == "Moderate Assignment"
    ).sum()
)

strong_assignment_count = int(
    (
        assignment_confidence_category
        == "Strong Assignment"
    ).sum()
)

assignment_condition_summary = pd.DataFrame({
    "Condition": [
        "Negative Silhouette",
        (
            "Relative Centroid Margin "
            f"< {BOUNDARY_MARGIN_THRESHOLD:.2f}"
        ),
        "Both Boundary Conditions",
        "Combined Boundary / Ambiguous",
        "Moderate Assignment",
        "Strong Assignment"
    ],
    "Customer_Count": [
        negative_silhouette_count,
        low_centroid_margin_count,
        both_boundary_conditions_count,
        combined_boundary_count,
        moderate_assignment_count,
        strong_assignment_count
    ]
})

assignment_condition_summary[
    "Customer_Percentage"
] = (
    assignment_condition_summary[
        "Customer_Count"
    ]
    / len(raw_clustered)
    * 100
)

confidence_category_summary = (
    raw_clustered[
        "Assignment_Confidence_Category"
    ]
    .value_counts()
    .reindex(
        [
            "Strong Assignment",
            "Moderate Assignment",
            "Boundary / Ambiguous Assignment"
        ],
        fill_value=0
    )
    .rename_axis(
        "Assignment_Confidence_Category"
    )
    .reset_index(
        name="Customer_Count"
    )
)

confidence_category_summary[
    "Customer_Percentage"
] = (
    confidence_category_summary[
        "Customer_Count"
    ]
    / len(raw_clustered)
    * 100
)

print("=" * 80)
print("OVERALL CUSTOMER-ASSIGNMENT CONFIDENCE")
print("=" * 80)

display(
    assignment_condition_summary.round(4)
)

display(
    confidence_category_summary.round(4)
)

# SEGMENT-LEVEL CONFIDENCE SUMMARY

confidence_by_segment = (
    raw_clustered
    .groupby(
        [
            "Cluster",
            "Segment_Name"
        ]
    )
    .agg(
        Customer_Count=(
            "CustomerID",
            "count"
        ),
        Mean_Silhouette=(
            "Silhouette_Value",
            "mean"
        ),
        Median_Silhouette=(
            "Silhouette_Value",
            "median"
        ),
        Mean_Relative_Centroid_Margin=(
            "Relative_Centroid_Margin",
            "mean"
        ),
        Median_Relative_Centroid_Margin=(
            "Relative_Centroid_Margin",
            "median"
        ),
        Negative_Silhouette_Customers=(
            "Negative_Silhouette_Flag",
            "sum"
        ),
        Low_Centroid_Margin_Customers=(
            "Low_Centroid_Margin_Flag",
            "sum"
        ),
        Boundary_Ambiguous_Customers=(
            "Combined_Boundary_Flag",
            "sum"
        )
    )
    .reset_index()
)

percentage_pairs = [
    (
        "Negative_Silhouette_Customers",
        "Negative_Silhouette_Percentage"
    ),
    (
        "Low_Centroid_Margin_Customers",
        "Low_Centroid_Margin_Percentage"
    ),
    (
        "Boundary_Ambiguous_Customers",
        "Boundary_Ambiguous_Percentage"
    )
]

for count_column, percentage_column in percentage_pairs:
    confidence_by_segment[
        percentage_column
    ] = (
        confidence_by_segment[
            count_column
        ]
        / confidence_by_segment[
            "Customer_Count"
        ]
        * 100
    )

segment_confidence_counts = (
    raw_clustered
    .groupby(
        [
            "Cluster",
            "Segment_Name",
            "Assignment_Confidence_Category"
        ]
    )
    .size()
    .unstack(
        fill_value=0
    )
    .reset_index()
)

confidence_by_segment = (
    confidence_by_segment
    .merge(
        segment_confidence_counts,
        on=[
            "Cluster",
            "Segment_Name"
        ],
        how="left",
        validate="one_to_one"
    )
    .round(4)
)

print("=" * 80)
print("CUSTOMER-ASSIGNMENT CONFIDENCE BY SEGMENT")
print("=" * 80)

display(confidence_by_segment)
```

### 26. Identify each customer's second-closest segment

The second-nearest centroid is treated as the most plausible alternative
K-Means segment.

This does not prove that a customer was incorrectly assigned. It shows which
segment is closest when the customer lies near a decision boundary.


```python
# CUSTOMER-LEVEL CONFIDENCE AND SEGMENT-OVERLAP DATA

customer_assignment_confidence = (
    raw_clustered
    .copy()
)

boundary_ambiguous_customers = (
    customer_assignment_confidence[
        customer_assignment_confidence[
            "Combined_Boundary_Flag"
        ]
    ]
    .copy()
    .sort_values(
        [
            "Relative_Centroid_Margin",
            "Silhouette_Value"
        ],
        ascending=[
            True,
            True
        ]
    )
    .reset_index(drop=True)
)

assigned_vs_alternative_matrix = pd.crosstab(
    customer_assignment_confidence[
        "Segment_Name"
    ],
    customer_assignment_confidence[
        "Alternative_Segment"
    ],
    margins=True,
    margins_name="Total"
)

boundary_overlap_matrix = pd.crosstab(
    boundary_ambiguous_customers[
        "Segment_Name"
    ],
    boundary_ambiguous_customers[
        "Alternative_Segment"
    ],
    margins=True,
    margins_name="Total"
)

segment_overlap_summary = (
    boundary_ambiguous_customers
    .groupby(
        [
            "Cluster",
            "Segment_Name",
            "Alternative_Cluster",
            "Alternative_Segment"
        ]
    )
    .agg(
        Boundary_Customer_Count=(
            "CustomerID",
            "count"
        ),
        Mean_Silhouette=(
            "Silhouette_Value",
            "mean"
        ),
        Mean_Relative_Centroid_Margin=(
            "Relative_Centroid_Margin",
            "mean"
        ),
        Median_Relative_Centroid_Margin=(
            "Relative_Centroid_Margin",
            "median"
        )
    )
    .reset_index()
)

assigned_segment_boundary_totals = (
    boundary_ambiguous_customers
    .groupby(
        [
            "Cluster",
            "Segment_Name"
        ]
    )
    .size()
    .reset_index(
        name="Assigned_Segment_Boundary_Total"
    )
)

segment_overlap_summary = (
    segment_overlap_summary
    .merge(
        assigned_segment_boundary_totals,
        on=[
            "Cluster",
            "Segment_Name"
        ],
        how="left",
        validate="many_to_one"
    )
)

segment_overlap_summary[
    "Percentage_of_Assigned_Segment_Boundary_Customers"
] = (
    segment_overlap_summary[
        "Boundary_Customer_Count"
    ]
    / segment_overlap_summary[
        "Assigned_Segment_Boundary_Total"
    ]
    * 100
)

segment_overlap_summary = (
    segment_overlap_summary
    .sort_values(
        [
            "Boundary_Customer_Count",
            "Segment_Name"
        ],
        ascending=[
            False,
            True
        ]
    )
    .reset_index(drop=True)
    .round(4)
)

print("=" * 80)
print("BOUNDARY CUSTOMER SEGMENT OVERLAP")
print("=" * 80)

display(
    segment_overlap_summary.head(20)
)

print("\nAll customers: assigned versus alternative segment")
display(assigned_vs_alternative_matrix)

print("\nBoundary customers only")
display(boundary_overlap_matrix)
```

### 27. Test sensitivity to the centroid-margin threshold

There is no universal centroid-margin threshold for K-Means. The table below
shows how the number of review cases changes when the margin condition is made
more or less strict.

The negative-silhouette condition remains unchanged in every scenario.


```python
# CENTROID-MARGIN THRESHOLD SENSITIVITY

margin_thresholds = [
    0.025,
    0.050,
    0.075,
    0.100,
    0.125,
    0.150,
    0.200
]

threshold_sensitivity_rows = []

for threshold in margin_thresholds:
    threshold_low_margin_flag = (
        relative_centroid_margin
        < threshold
    )

    threshold_combined_flag = (
        negative_silhouette_flag
        | threshold_low_margin_flag
    )

    threshold_sensitivity_rows.append({
        "Relative_Centroid_Margin_Threshold": threshold,
        "Low_Margin_Customers": int(
            threshold_low_margin_flag.sum()
        ),
        "Low_Margin_Percentage": float(
            threshold_low_margin_flag.mean()
            * 100
        ),
        "Combined_Boundary_Customers": int(
            threshold_combined_flag.sum()
        ),
        "Combined_Boundary_Percentage": float(
            threshold_combined_flag.mean()
            * 100
        )
    })

confidence_threshold_sensitivity = (
    pd.DataFrame(
        threshold_sensitivity_rows
    )
    .round(4)
)

display(confidence_threshold_sensitivity)
```

## Phase 5 — Final frozen-model stability testing

The final solution is tested against changes in random initialisation and customer sampling.


### 30. Test stability across random seeds

A stable solution should not change greatly because of a different centroid starting point.The K=5 model is fitted repeatedly using different random seeds and each result is compared with the final labels using Adjusted Rand Index.



```python
seed_results = []

for seed in range(1, 31):
    seed_model = KMeans(
        n_clusters=FINAL_K,
        init="k-means++",
        n_init=N_INIT,
        max_iter=MAX_ITER,
        random_state=seed,
        algorithm="lloyd"
    )

    seed_labels = seed_model.fit_predict(X)

    seed_results.append({
        "Random_State": seed,
        "Adjusted_Rand_Index": adjusted_rand_score(
            final_labels,
            seed_labels
        ),
        "Inertia": seed_model.inertia_,
        "Iterations": seed_model.n_iter_
    })

seed_stability = pd.DataFrame(seed_results)

seed_stability_summary = pd.DataFrame({
    "Metric": [
        "Runs",
        "Mean ARI",
        "Median ARI",
        "Minimum ARI",
        "Maximum ARI",
        "ARI Standard Deviation",
        "Perfect Matches"
    ],
    "Value": [
        len(seed_stability),
        seed_stability["Adjusted_Rand_Index"].mean(),
        seed_stability["Adjusted_Rand_Index"].median(),
        seed_stability["Adjusted_Rand_Index"].min(),
        seed_stability["Adjusted_Rand_Index"].max(),
        seed_stability["Adjusted_Rand_Index"].std(),
        (seed_stability["Adjusted_Rand_Index"] == 1).sum()
    ]
})

print("=" * 80)
print("RANDOM-SEED STABILITY")
print("=" * 80)

display(seed_stability_summary)

seed_stability.to_csv(
    stability_dir / "final_k5_random_seed_runs.csv",
    index=False
)

seed_stability_summary.to_csv(
    stability_dir / "final_k5_random_seed_summary.csv",
    index=False
)
```

### 31. Test stability using repeated customer subsamples

The solution should also remain reasonably similar when part of the customer sample changes. Fifty models are fitted on random samples containing 80 percent of customers and compared with the final labels for those customers.


```python
SUBSAMPLE_RUNS = 50
SUBSAMPLE_FRACTION = 0.80
SUBSAMPLE_RANDOM_STATE = 2026

random_generator = np.random.default_rng(
    SUBSAMPLE_RANDOM_STATE
)

sample_size = int(
    len(X) * SUBSAMPLE_FRACTION
)

subsample_results = []

for run_number in range(1, SUBSAMPLE_RUNS + 1):
    sampled_positions = random_generator.choice(
        len(X),
        size=sample_size,
        replace=False
    )

    sampled_positions = np.sort(sampled_positions)

    X_subsample = X.iloc[sampled_positions]
    baseline_sample_labels = final_labels[sampled_positions]

    subsample_model = KMeans(
        n_clusters=FINAL_K,
        init="k-means++",
        n_init=N_INIT,
        max_iter=MAX_ITER,
        random_state=SUBSAMPLE_RANDOM_STATE + run_number,
        algorithm="lloyd"
    )

    subsample_labels = subsample_model.fit_predict(
        X_subsample
    )

    subsample_results.append({
        "Run": run_number,
        "Sample_Size": sample_size,
        "Sample_Fraction": SUBSAMPLE_FRACTION,
        "Adjusted_Rand_Index": adjusted_rand_score(
            baseline_sample_labels,
            subsample_labels
        ),
        "Inertia": subsample_model.inertia_,
        "Iterations": subsample_model.n_iter_
    })

subsample_stability = pd.DataFrame(
    subsample_results
)

subsample_stability_summary = pd.DataFrame({
    "Metric": [
        "Runs",
        "Sample Fraction",
        "Sample Size",
        "Mean ARI",
        "Median ARI",
        "Minimum ARI",
        "Maximum ARI",
        "ARI Standard Deviation"
    ],
    "Value": [
        SUBSAMPLE_RUNS,
        SUBSAMPLE_FRACTION,
        sample_size,
        subsample_stability["Adjusted_Rand_Index"].mean(),
        subsample_stability["Adjusted_Rand_Index"].median(),
        subsample_stability["Adjusted_Rand_Index"].min(),
        subsample_stability["Adjusted_Rand_Index"].max(),
        subsample_stability["Adjusted_Rand_Index"].std()
    ]
})

print("=" * 80)
print("SUBSAMPLING STABILITY")
print("=" * 80)

display(subsample_stability_summary)

subsample_stability.to_csv(
    stability_dir / "final_k5_subsampling_runs.csv",
    index=False
)

subsample_stability_summary.to_csv(
    stability_dir / "final_k5_subsampling_summary.csv",
    index=False
)
```
